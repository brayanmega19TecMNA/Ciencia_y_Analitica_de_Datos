2. Determina el tipo de datos que tienen las columnas.
- Cambia la columna `medicion` a datetime utilizando el formato: `%d/%m/%y %H:%M`
- Obtén dos columnas adicionales separando `medicion` en `fecha` y `hora`

4. Imprime nuevamente las cantidad de valores únicos por columna.

- La cantidad de combinaciones únicas de fecha y hora no coincide con el número total de mediciones. Aunque no haya valores NaN, esto no garantiza que no se hayan omitido registros. Identifica las fechas en las que faltan mediciones.
- Una vez concluido el análisis de mediciones faltantes, elimina todos los registros que no correspondan al año 2020 y la columna `medicion`.


```python
cantidadValoresUnicos = weather_df.nunique(dropna=False)

print("4. Cantidad de valores únicos por columna:")
print(cantidadValoresUnicos)


print("¿Por qué son 144 horas diferentes?")
medicionesPorHora = 60 // 10
horariosEsperadosPorDia = 24 * medicionesPorHora

print(f"Mediciones por hora: {medicionesPorHora}")
print(f"Horarios diferentes por día: {horariosEsperadosPorDia} (6 mediciones por hora, multiplicadas por las 24 horas del dia)")

print("\nLa cantidad de combinaciones únicas de fecha y hora no coincide con el número total de mediciones. Aunque no haya valores NaN, esto no garantiza que no se hayan omitido registros. Identifica las fechas en las que faltan mediciones.")
cantidadFechas = weather_df["fecha"].nunique()
cantidadHoras = weather_df["hora"].nunique()

combinacionesEsperadas = cantidadFechas * cantidadHoras
totalMediciones = weather_df.shape[0]
medicionesFaltantes = combinacionesEsperadas - totalMediciones
print("Mediciones faltantes", medicionesFaltantes)


print("Una vez concluido el análisis de mediciones faltantes, elimina todos los registros que no correspondan al año 2020 y la columna medicion.")

registrosDel2020 = weather_df["medicion"].dt.year == 2020
weather_df = weather_df[registrosDel2020].copy()
weather_df = weather_df.drop(columns=["medicion"])
weather_df
```

5. Obtén un dataframe `indicadores_diarios` que sólo almacene el mayor valor de las mediciones por fecha para responder:
- ¿Cuál fue la máxima precipitación y a qué fecha pertenece?
- ¿Cuáles fueron los tres días con las velocidades de viento más altas registradas?
- ¿Cuántos días rebasaron el 90% de humedad relativa y los 30° celsius?

```python
print("Obtén un dataframe indicadores_diarios que sólo almacene el mayor valor de las mediciones por fecha para responder:")

indicadores_diarios = weather_df.groupby(
    "fecha",
    as_index=False
).max(numeric_only=True)

print(indicadores_diarios)


print("¿Cuál fue la máxima precipitación y a qué fecha pertenece?")
maximaPrecipitacion = indicadores_diarios[
    "precipitacion_total"
].max()

fechaMaximaPrecipitacion = indicadores_diarios.loc[
    indicadores_diarios["precipitacion_total"].idxmax(),
    "fecha"
]

print(f"Máxima precipitación: {maximaPrecipitacion}")
print(f"Fecha correspondiente: {fechaMaximaPrecipitacion}")

print("¿Cuáles fueron los tres días con las velocidades de viento más altas registradas?")

diasMayorVelocidadViento = indicadores_diarios.nlargest(
    3,
    "velocidad_viento"
)[["fecha", "velocidad_viento"]]

print("Tres días con las velocidades de viento más altas:")
print(diasMayorVelocidadViento.to_string(index=False))
```

8. Utiliza el dataframe de indicadores diarios para crear un nuevo dataframe `indicadores_mensuales`. Para ello:
- Crea una nueva columna `mes` a partir de la fecha que se encuentra en el índice.
- Haciendo uso de `groupby()`, calcula los promedios mensuales y conserva únicamente las columnas `humedad_relativa`, `temperatura_celsius` y `velocidad_viento` en el nuevo dataframe.

```python
print("Crea una nueva columna mes a partir de la fecha que se encuentra en el índice.")

indicadores_diarios = indicadores_diarios.set_index("fecha")

indicadores_diarios.index = pd.to_datetime(
    indicadores_diarios.index
)

indicadores_diarios["mes"] = indicadores_diarios.index.month

print("Haciendo uso de groupby(), calcula los promedios mensuales y conserva únicamente las columnas humedad_relativa, temperatura_celsius y velocidad_viento en el nuevo dataframe.")

columnasPromediosMensuales = [
    "humedad_relativa",
    "temperatura_celsius",
    "velocidad_viento"
]

indicadores_mensuales = indicadores_diarios.groupby(
    "mes"
)[columnasPromediosMensuales].mean()

print("Promedios mensuales:")
print(indicadores_mensuales.to_string())
```

9. Construye el mismo dataframe mensual, pero esta vez utilizando la función `pivot_table()`.
- Añade la columna `indice_calor` usando la fórmula de Thom (*Temperature - Humidity Index*, TDI):
> `temperatura_celsius + 0.33 x humedad_relativa - 0.7 x velocidad_viento - 4`

```python
print("Construye el mismo dataframe mensual, pero esta vez utilizando la función pivot_table().")
indicadores_mensuales_pivot = indicadores_diarios.pivot_table(
    index="mes",
    values=[
        "humedad_relativa",
        "temperatura_celsius",
        "velocidad_viento"
    ],
    aggfunc="mean"
)
#En este punto empiezo a reaizar un calculo, para tener el indice de calor

indicadores_mensuales_pivot["indice_calor"] = (
    indicadores_mensuales_pivot["temperatura_celsius"]
    + 0.33 * indicadores_mensuales_pivot["humedad_relativa"]
    - 0.7 * indicadores_mensuales_pivot["velocidad_viento"]
    - 4
)

print("\nIndicadores mensuales con índice de calor:")
print(indicadores_mensuales_pivot.to_string())
```

10. Recuerda que de las cuatro funciones estudiadas para manipular la estructura del dataframe:
- `melt/pivot` hacen las transformaciones de manera controlada, es decir puedes
definir por medio de sus parámetros qué variables quedarán como índices, columnas y valores.
- `stack/unstack` la conversión se aplica siempre sobre los niveles inferiores de index/columns.
- Convierte el dataframe del ejercicio anterior a formato largo usando: `melt()` y `stack()` y comenta las diferencias.

```python
columnasFormatoLargo = [
    "humedad_relativa",
    "temperatura_celsius",
    "velocidad_viento",
    "indice_calor"
]

indicadores_largo_melt = (
    indicadores_mensuales_pivot
    .reset_index()
    .melt(
        id_vars="mes",
        value_vars=columnasFormatoLargo,
        var_name="indicador",
        value_name="valor"
    )
)

print("Formato largo utilizando melt:")
print(indicadores_largo_melt.to_string(index=False))


print("\nFormato largo utilizando stack:")

indicadores_largo_stack = (
    indicadores_mensuales_pivot
    .stack()
    .rename_axis(index=["mes", "indicador"])
    .reset_index(name="valor")
)

print(indicadores_largo_stack.to_string(index=False))
```
